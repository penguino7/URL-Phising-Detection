```python
import sys
sys.path.append("..")

import pandas as pd
from Src.Core.url_parser import parse_url
from Src.Core.config import DATA_INTERIM_DIR
```

```python
sample_urls = [
    "https://sub.vietcombank.com.vn:8080/ebanking/login?token=abc#top",
    "https://vcb-login.pages.dev/auth",
    "http://scam-bank.top/verify",
    "http://192.168.1.1:8888/test",
    "shopee.vn/flash-sale"
]

results = [parse_url(u) for u in sample_urls]

# Hiển thị dạng bảng trực quan
df_sample = pd.DataFrame([
    {
        "raw_url": r.raw_url,
        "scheme": r.scheme,
        "host": r.host,
        "port": r.port,
        "subdomain": r.subdomain,
        "domain": r.domain,
        "suffix (eTLD)": r.suffix,
        "main_tld": r.main_tld,
        "registered_domain": r.registered_domain,
        "is_private_suffix": r.is_private_suffix
    }
    for r in results
])

df_sample
```

```python
df_full = pd.read_csv(DATA_INTERIM_DIR / "urls_cleaned.csv")

records = []
for url, label in zip(df_full["url"], df_full["label"]):
    p = parse_url(url)
    records.append({
        "url": url,
        "label": label,
        "scheme": p.scheme,
        "netloc": p.netloc,
        "host": p.host,
        "port": p.port,
        "path": p.path,
        "query": p.query,
        "subdomain": p.subdomain,
        "domain": p.domain,
        "suffix": p.suffix,
        "registered_domain": p.registered_domain,
        "main_tld": p.main_tld,
        "is_private_suffix": p.is_private_suffix,
    })

df_parsed = pd.DataFrame(records)
df_parsed.to_csv(DATA_INTERIM_DIR / "urls_parsed.csv", index=False)
print("Số lượng dòng đã xuất:", len(df_parsed))
```
